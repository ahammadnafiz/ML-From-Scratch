```python
import os
from dotenv import load_dotenv
# Paid API keys
# from langchain_openai import ChatOpenAI
from langchain_google_genai import ChatGoogleGenerativeAI
from langchain_core.prompts import ChatPromptTemplate
from langchain_core.output_parsers import StrOutputParser
```

```python
# Load the .env file
load_dotenv()
```

```python
# os.environ["OPENAI_API_KEY"]=os.getenv("OPENAI_API_KEY")
```

```python
api_key = os.getenv("GOOGLE_API_KEY")
langchain_api = os.getenv("LANGCHAIN_API_KEY")
os.environ["LANGCHAIN_TRACING_V2"] = "true"
# print(api_key)
```

```python
template = ChatPromptTemplate([
    ("system", "You are a helpful AI bot. Your name is Carl."),
    ("human", "{input}"),
])
```

```python
# llm=ChatOpenAI(model="gpt-3.5-turbo")
# For Local Model
# llm=Ollama(model="llama2")
llm = ChatGoogleGenerativeAI(api_key=api_key, model="gemini-1.5-pro")
output_parser = StrOutputParser()
chain = template|llm|output_parser
```

```python
response = chain.invoke({"input": "How do I fix my computer?"})
print(response)
```

```python
response = chain.invoke({"input": "How to swap two variables in Python?"})
print(response)
```
