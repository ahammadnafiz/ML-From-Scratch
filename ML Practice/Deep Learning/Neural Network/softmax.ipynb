```python
import numpy as np
import matplotlib.pyplot as plt
from mpl_toolkits.mplot3d import Axes3D

# Define softmax function
def softmax(z):
    z = z - np.max(z, axis=-1, keepdims=True)  # stability
    exp_z = np.exp(z)
    return exp_z / np.sum(exp_z, axis=-1, keepdims=True)

# Create a 2D grid of values
x = np.linspace(-5, 5, 100)
y = np.linspace(-5, 5, 100)
X, Y = np.meshgrid(x, y)

# Apply softmax to each pair (X, Y)
Z1 = softmax(np.stack([X, Y], axis=-1))[..., 0]  # softmax output for 1st element
Z2 = softmax(np.stack([X, Y], axis=-1))[..., 1]  # softmax output for 2nd element

# Plot in 2D (contour)
plt.figure(figsize=(12, 5))

plt.subplot(1, 2, 1)
cp = plt.contourf(X, Y, Z1, levels=50, cmap='viridis')
plt.colorbar(cp)
plt.title('Softmax Component 1')
plt.xlabel('x')
plt.ylabel('y')

plt.subplot(1, 2, 2)
cp = plt.contourf(X, Y, Z2, levels=50, cmap='plasma')
plt.colorbar(cp)
plt.title('Softmax Component 2')
plt.xlabel('x')
plt.ylabel('y')

plt.tight_layout()
plt.show()

# 3D plot
fig = plt.figure(figsize=(12, 5))

ax = fig.add_subplot(121, projection='3d')
ax.plot_surface(X, Y, Z1, cmap='viridis')
ax.set_title('Softmax Component 1 (3D)')
ax.set_xlabel('x')
ax.set_ylabel('y')
ax.set_zlabel('softmax(x)')

ax = fig.add_subplot(122, projection='3d')
ax.plot_surface(X, Y, Z2, cmap='plasma')
ax.set_title('Softmax Component 2 (3D)')
ax.set_xlabel('x')
ax.set_ylabel('y')
ax.set_zlabel('softmax(y)')

plt.tight_layout()
plt.show()
```
