### Tensor Shapes

```python
tensor = torch.tensor([
    [1, 2],
    [3, 4]
], dtype=torch.float32) # creates a tensor with specific values and type
```

```python
tensor.shape # returns the shape of the tensor
```

```python
# empty_like, ones_like, zeros_like
# creates a tensor with the same shape as another tensor
# this is useful when you want to create a tensor with the same shape as an existing tensor

x = torch.empty_like(tensor) # creates an empty tensor with the same shape as tensor
print(x)
y = torch.ones_like(tensor) # creates a tensor filled with ones with the same shape as tensor
print(y)
z = torch.zeros_like(tensor) # creates a tensor filled with zeros with the same shape as tensor
print(z)
i = torch.rand_like(tensor) # creates a tensor with random values with the same shape as tensor
print(i)
```

| **Data Type**             | **Dtype**         | **Description**                                                                                                                                                                |
|---------------------------|-------------------|--------------------------------------------------------------------------------------------------------------------------------------------------------------------------------|
| **32-bit Floating Point** | `torch.float32`   | Standard floating-point type used for most deep learning tasks. Provides a balance between precision and memory usage.                                                         |
| **64-bit Floating Point** | `torch.float64`   | Double-precision floating point. Useful for high-precision numerical tasks but uses more memory.                                                                               |
| **16-bit Floating Point** | `torch.float16`   | Half-precision floating point. Commonly used in mixed-precision training to reduce memory and computational overhead on modern GPUs.                                            |
| **BFloat16**              | `torch.bfloat16`  | Brain floating-point format with reduced precision compared to `float16`. Used in mixed-precision training, especially on TPUs.                                                |
| **8-bit Floating Point**  | `torch.float8`    | Ultra-low-precision floating point. Used for experimental applications and extreme memory-constrained environments (less common).                                               |
| **8-bit Integer**         | `torch.int8`      | 8-bit signed integer. Used for quantized models to save memory and computation in inference.                                                                                   |
| **16-bit Integer**        | `torch.int16`     | 16-bit signed integer. Useful for special numerical tasks requiring intermediate precision.                                                                                    |
| **32-bit Integer**        | `torch.int32`     | Standard signed integer type. Commonly used for indexing and general-purpose numerical tasks.                                                                                  |
| **64-bit Integer**        | `torch.int64`     | Long integer type. Often used for large indexing arrays or for tasks involving large numbers.                                                                                  |
| **8-bit Unsigned Integer**| `torch.uint8`     | 8-bit unsigned integer. Commonly used for image data (e.g., pixel values between 0 and 255).                                                                                    |
| **Boolean**               | `torch.bool`      | Boolean type, stores `True` or `False` values. Often used for masks in logical operations.                                                                                      |
| **Complex 64**            | `torch.complex64` | Complex number type with 32-bit real and 32-bit imaginary parts. Used for scientific and signal processing tasks.                                                               |
| **Complex 128**           | `torch.complex128`| Complex number type with 64-bit real and 64-bit imaginary parts. Offers higher precision but uses more memory.                                                                 |
| **Quantized Integer**     | `torch.qint8`     | Quantized signed 8-bit integer. Used in quantized models for efficient inference.                                                                                              |
| **Quantized Unsigned Integer** | `torch.quint8` | Quantized unsigned 8-bit integer. Often used for quantized tensors in image-related tasks.                                                                                     |


### Mathematical Operations

```python
x = torch.tensor([
    [1, 2],
    [3, 4]
], dtype=torch.float32) # creates a tensor with specific values and type
print(x)
```

```python
# addition, subtraction, multiplication, division, int_division, modulus, exponentiation
print(x + 2) # adds 2 to each element of the tensor
print(x - 2) # subtracts 2 from each element of the tensor
print(x * 2) # multiplies each element of the tensor by 2
print(x / 2) # divides each element of the tensor by 2
print(x // 2) # integer division of each element of the tensor by 2
print(x % 2) # modulus of each element of the tensor by 2
print(x ** 2) # exponentiation of each element of the tensor by 2
print(torch.clamp(x, 2, 3)) # clamps the values of the tensor between 2 and 3

# using in-place operations
x.add_(2) # adds 2 to each element of the tensor in-place
print(x)
x.sub_(2) # subtracts 2 from each element of the tensor in-place
print(x)
x.mul_(2) # multiplies each element of the tensor by 2 in-place
print(x)
x.div_(2) # divides each element of the tensor by 2 in-place
print(x)
x.int() # converts the tensor to int
print(x)

# using item()
# converts a single-element tensor to a Python number
x = torch.tensor(3.14)
print(x.item()) # prints 3.14
```

### Reduction Operations

```python
tensor = torch.randint(size=(3, 4), low=0, high=10, dtype=torch.float64) # creates a tensor with random integer values
```

```python
tensor
```

```python
#summing
#sum of all elements
print(tensor.sum()) # prints the sum of all elements of the tensor
#sum of each column
print(tensor.sum(dim=0)) # prints the sum of each column of the tensor
#sum of each row
print(tensor.sum(dim=1)) # prints the sum of each row of the tensor
#sum of each element
print(tensor.sum(dim=-1)) # prints the sum of each element of the tensor
print(tensor.sum(dim=0, keepdim=True)) # prints the sum of each element of the tensor with keepdim=True
print(tensor.sum(dim=1, keepdim=True)) # prints the sum of each element of the tensor with keepdim=True
print(tensor.sum(dim=-1, keepdim=True)) # prints the sum of each element of the tensor with keepdim=True
print(tensor.sum(dim=0, keepdim=False)) # prints the sum of each element of the tensor with keepdim=False
print(tensor.sum(dim=1, keepdim=False)) # prints the sum of each element of the tensor with keepdim=False
print(tensor.sum(dim=-1, keepdim=False)) # prints the sum of each element of the tensor with keepdim=False
print(tensor.sum(dim=0, keepdim=True)) # prints the sum of each element of the tensor with keepdim=True
print(tensor.sum(dim=1, keepdim=True)) # prints the sum of each element of the tensor with keepdim=True
```

```python
# Mean
# mean of all elements
print(tensor.mean()) # prints the mean of all elements of the tensor
```

```python
# max, min
# max of all elements
print(tensor.max()) # prints the max of all elements of the tensor
print(tensor.min()) # prints the min of all elements of the tensor
```

```python
# argmax, argmin
# index of max of all elements
print(tensor.argmax()) # prints the index of the max of all elements of the tensor
print(tensor.argmin()) # prints the index of the min of all elements of the tensor
```

### Comparision Opearations

```python
x = torch.randint(size=(3, 4), low=0, high=10, dtype=torch.float64) # creates a tensor with random integer values
print(x)
y = torch.randint(size=(3, 4), low=0, high=10, dtype=torch.float64) # creates a tensor with random integer values
print(y)
```

```python
# greater than, less than, greater than or equal to, less than or equal to
print(x > y) # prints True if x is greater than y
print(x < y) # prints True if x is less than y
print(x >= y) # prints True if x is greater than or equal to y
print(x <= y) # prints True if x is less than or equal to y
print(x == y) # prints True if x is equal to y
print(x != y) # prints True if x is not equal to y
print(x.eq(y)) # prints True if x is equal to y
print(x.ne(y)) # prints True if x is not equal to y
print(x.gt(y)) # prints True if x is greater than y
print(x.lt(y)) # prints True if x is less than y
print(x.ge(y)) # prints True if x is greater than or equal to y
print(x.le(y)) # prints True if x is less than or equal to y
```

```python
# log, exp, sqrt, abs, sigmoid, tanh, relu
print(torch.log(x)) # prints the natural logarithm of x
print(torch.exp(x)) # prints the exponential of x
print(torch.sqrt(x)) # prints the square root of x
print(torch.abs(x)) # prints the absolute value of x
print(torch.sigmoid(x)) # prints the sigmoid of x
print(torch.tanh(x)) # prints the hyperbolic tangent of x
print(torch.relu(x)) # prints the rectified linear unit of x
print(torch.softmax(x, dim=0)) # prints the softmax of x along dimension 0
print(torch.softmax(x, dim=1)) # prints the softmax of x along dimension 1
print(torch.softmax(x, dim=-1)) # prints the softmax of x along the last dimension
```
